
## 🔹 NP1–NP10: Array Conversions & Basic Integration


```python
# NP1. Create a NumPy array and convert it to a Pandas Series.
#Hint:** `np.array([...])` creates the array. Pass it directly to `pd.Series(array, index=[...], name="...")`.

import numpy as np
import pandas as pd

arr = np.array([10, 20, 30, 40, 50])
s = pd.Series(arr, index=['A', 'B', 'C', 'D', 'E'], name='Numbers')
print(s)
```

```python
# NP2. Convert a 2D NumPy array to a Pandas DataFrame.
#Hint:** `pd.DataFrame(np_2d_array, columns=[...])` — shape `(n, m)` gives `n` rows and `m` columns.

arr = np.array([
    [101, 'Snehal', 85],
    [102, 'Komal', 90],
    [103, 'Priya', 88]
])
df = pd.DataFrame(arr, columns=['ID', 'Name', 'Marks'])
print(df)
```

```python
# NP3. Generate a random dataset using NumPy and store it in a DataFrame.
#Hint:** Use `np.random.seed()` for reproducibility. Try `np.random.randn(n, m)` for normal distribution or `np.random.randint(lo, hi, n)` for integers.

np.random.seed(42)
data = np.random.randn(5, 3)
df = pd.DataFrame(data, columns=['Feature1', 'Feature2', 'Feature3'])
print(df)
```

```python
# NP4. Perform element-wise operations using NumPy and Pandas.
#Hint:** You can apply NumPy ufuncs directly on a Pandas Series:
#  `np.sqrt(df["col"])`. The result is a Series.

df = pd.DataFrame({
    'Numbers': [4, 9, 16, 25, 36]
})
df['Square_Root'] = np.sqrt(df['Numbers'])
print(df)
```

```python
# NP5. Apply `np.mean` on DataFrame columns.
#Hint:** `np.mean(df["col"])` works the same as `df["col"].mean()`. Apply to the `.values` array for pure NumPy speed: `np.mean(df["col"].values)`.

df = pd.DataFrame({
    'Marks': [85, 90, 78, 92, 88]
})
mean_marks = np.mean(df['Marks'])
print(mean_marks)
```

```python
# NP6. Replace values using `np.where`.
#Hint:** `np.where(condition, value_if_true, value_if_false)` — assign the result back to a column: `df["col"] = np.where(df["col"] > 50, "High", "Low")`.

df = pd.DataFrame({
    'Marks': [45, 78, 52, 30, 90]
})
df['Result'] = np.where(df['Marks'] > 50, 'High', 'Low')
print(df)
```

```python
# NP7. Apply logarithmic transformation using `np.log`.
#Hint:** `df["col_log"] = np.log(df["col"])`. Note: `np.log(0)` returns `-inf`. Use `np.log1p(x)` which computes `log(1 + x)` to safely handle zeros.

df = pd.DataFrame({
    'Sales': [10, 20, 50, 100, 200]
})
df['Sales_Log'] = np.log(df['Sales'])
print(df)
```

```python
# NP8. Apply square root transformation using `np.sqrt`.
#Hint:** `df["col_sqrt"] = np.sqrt(df["col"])`. Values must be non-negative; use `np.clip(df["col"], 0, None)` first if negatives are possible.

df = pd.DataFrame({
    'Marks': [45, 78, 52, 30, 90]
})
df['Marks_Sqrt'] = np.sqrt(df['Marks'])
print(df)
```

```python
# NP9. Use NumPy functions inside `apply()`.
#Hint:** `df["col"].apply(lambda x: np.log(x))` — or pass the NumPy function directly: `df["col"].apply(np.log)`.

df['Marks_Log'] = df['Marks'].apply(np.log)
print(df)
```

```python
# NP10. Perform broadcasting in a DataFrame using NumPy.
#Hint:** Extract `.values` (a 2D NumPy array) and multiply by a weights array: 
# `df[cols].values @ weights_array`. 
# The result can be assigned back as a new column.

weights = np.array([0.3, 0.7])
df['Weighted_Score'] = df[['Marks', 'Marks_Sqrt']].values @ weights
print(df)

# NumPy automatically broadcasts the weights across all
#  rows and calculates a weighted score for each row
```

```python
# NP14. Flatten DataFrame values to a 1D NumPy array.
#Hint:** `df[["col1","col2"]].values.flatten()` — 
# the .flatten()`method converts the 2D array to 1D row-by-row.

df = pd.DataFrame({
    'Marks': [45, 78, 52, 30, 90]
})
df['Marks_Sqrt'] = np.sqrt(df['Marks'])
flat_array = df[['Marks', 'Marks_Sqrt']].values.flatten()
print(flat_array)
```

```python
# NP15. Reshape a NumPy array before creating a DataFrame.
#Hint:** `np.arange(1, 25).reshape(6, 4)` creates a 6×4 array. Pass directly to `pd.DataFrame(arr, columns=[...])`.

arr = np.arange(1, 25).reshape(6, 4)
df = pd.DataFrame(arr, columns=['A', 'B', 'C', 'D'])
print(df)
```

```python
# NP16. Perform matrix multiplication using `np.dot`.
#Hint:** `np.dot(A, B)` or `A @ B`. Extract `.values` from DataFrames first, then wrap the result in a new `pd.DataFrame(result, columns=[...])`.

A = pd.DataFrame([[1, 2],
                  [3, 4]])

B = pd.DataFrame([[5, 6],
                  [7, 8]])
result = np.dot(A.values, B.values)
result_df = pd.DataFrame(result, columns=['Col1', 'Col2'])
print(result_df)
```

```python
# NP17. Normalize data using NumPy (Min-Max scaling).
#Hint:** `arr = df[cols].values` then
#  `(arr - arr.min(axis=0)) / (arr.max(axis=0) - 
# arr.min(axis=0))`. Use `axis=0` to operate column-wise.

df = pd.DataFrame({
    'Marks': [45, 78, 52, 30, 90]
})
df['Marks_Sqrt'] = np.sqrt(df['Marks']) # Calculate the missing 'Marks_Sqrt' column

arr = df[['Marks', 'Marks_Sqrt']].values #Extract the 2D NumPy array

# Min-Max Scaling
normalized = (arr - arr.min(axis=0)) / (arr.max(axis=0) - arr.min(axis=0))

# normalized array back into a cleanly labeled DataFrame
normalized_df = pd.DataFrame(
    normalized,
    columns=['Marks_Normalized', 'Marks_Sqrt_Normalized']
)
print(normalized_df)
```

```python
# NP18. Standardize data using Z-score with NumPy.
#Hint:** `z = (arr - arr.mean(axis=0)) / arr.std(axis=0)`. Wrap the result in a DataFrame using the original column names.

arr = df[['Marks', 'Marks_Sqrt']].values
z_score = (arr - arr.mean(axis=0)) / arr.std(axis=0)

z_score_df = pd.DataFrame(
    z_score,
    columns=['Marks_ZScore', 'Marks_Sqrt_ZScore']
)
print(z_score_df)
```

```python
df['Marks_Clipped'] = np.clip(df['Marks'].values, a_min=50, a_max=90)
print(df)
```

```python
# NP20. Handle missing values using NumPy.
#Hint:** Detect NaN with `np.isnan(arr)`. Fill using `np.nanmean(arr)` to compute the mean ignoring NaN, then replace: `arr[np.isnan(arr)] = np.nanmean(arr)`.

df.loc[2, 'Marks'] = np.nan
arr = df['Marks'].values
arr[np.isnan(arr)] = np.nanmean(arr)
df['Marks'] = arr
print(df)
```

```python
# NP21. Compare NumPy vs Pandas performance on a large array.
#Hint:** Use `time.time()` before and after each operation. 
# Apply the same computation (`arr * 2 + 1`) to a raw NumPy array and a Pandas Series.
#  NumPy is typically faster on pure numeric operations.
import time

arr = np.random.rand(1000000)
series = pd.Series(arr)

start = time.time()
result_np = arr * 2 + 1
end = time.time()

print("NumPy Time:", end - start)

start = time.time()
result_pd = series * 2 + 1
end = time.time()

print("Pandas Time:", end - start)

```

```python
# NP22. Calculate cumulative sum using NumPy.
#Hint:** `np.cumsum(df["col"].values)` — assign back: `df["cum_col"] = np.cumsum(df["col"])`.

df['Marks_Cumulative'] = np.cumsum(df['Marks'])
print(df)
```

```python
# NP23. Use `np.argmax` and `np.argmin`.
#Hint:** `np.argmax(df["col"].values)` returns the **integer position** 
# of the maximum. Use `df.iloc[idx]` or 
# `df.loc[idx]` to retrieve the full row.

max_idx = np.argmax(df['Marks'].values)
min_idx = np.argmin(df['Marks'].values)

print("Maximum Value Row:")
print(df.iloc[max_idx])

print("\nMinimum Value Row:")
print(df.iloc[min_idx])
```

```python
# NP24. Sort values using NumPy `np.argsort`.
#Hint:** `sorted_idx = np.argsort(df["col"].values)` gives indices that would sort the array. Use `df.iloc[sorted_idx]` to reorder the DataFrame.

sorted_idx = np.argsort(df['Marks'].values)
sorted_df = df.iloc[sorted_idx]
print(sorted_df)
```

```python
# NP25. Compute correlation using `np.corrcoef`.
#Hint:** `np.corrcoef([df["col1"].values, 
# df["col2"].values])` returns a 2×2 correlation matrix. 
# Wrap in a DataFrame for readability. 
# Compare with `df.corr()`.

corr_matrix = np.corrcoef([
    df['Marks'].values,
    df['Marks_Sqrt'].values
])
corr_df = pd.DataFrame(
    corr_matrix,
    index=['Marks', 'Marks_Sqrt'],
    columns=['Marks', 'Marks_Sqrt']
)
print(corr_df)
```

```python
# NP26. Detect NaN values using NumPy.
#Hint:** `np.isnan(arr)` returns a boolean array. Count NaN with `np.isnan(arr).sum()`. Compare with `pd.isnull(series)`.

arr = df['Marks'].values

nan_mask = np.isnan(arr)
nan_count = np.isnan(arr).sum()

print("NaN Mask:")
print(nan_mask)

print("\nNaN Count:")
print(nan_count)
```

```python
# NP27. Create boolean masks using NumPy.
#Hint:** `mask = (np.array(df["col1"]) > 80) & 
# (np.array(df["col2"]) < 40)` — use `df[mask]` to filter.

mask = (np.array(df['Marks']) > 80) & (np.array(df['Marks_Sqrt']) < 10)
filtered_df = df[mask]
print(filtered_df)
```

```python
# NP28. Apply logical operations (`np.logical_and`, `np.logical_or`, `np.logical_not`).
#Hint:** `np.logical_and(cond1, cond2)` is equivalent to `cond1 & cond2`. These functions work on NumPy arrays; convert Series with `.values` first.

cond1 = df['Marks'].values > 80
cond2 = df['Marks_Sqrt'].values < 10

and_mask = np.logical_and(cond1, cond2)
or_mask = np.logical_or(cond1, cond2)
not_mask = np.logical_not(cond1)

print("Logical AND:")
print(df[and_mask])

print("\nLogical OR:")
print(df[or_mask])

print("\nLogical NOT:")
print(df[not_mask])
```

```python
# NP29. Apply `np.select` for multi-condition assignment.
#Hint:** `df["Grade"] = np.select([df["Score"] >= 90, 
# df["Score"] >= 75], ["A","B"], default="C")` 
# — conditions are checked in order; the first match wins.

conditions = [
    df['Marks'] >= 90,
    df['Marks'] >= 75
]
choices = ['A', 'B']
df['Grade'] = np.select(conditions, choices, default='C')
print(df)
```

```python
# NP30. Multiply DataFrame columns element-wise using NumPy.
#Hint:** `np.multiply(df["col1"].values, df["col2"].values)`
#  is equivalent to `df["col1"] * df["col2"]` but uses NumPy internally.

df['Product'] = np.multiply(
    df['Marks'].values,
    df['Marks_Sqrt'].values
)
print(df)
```
