```python
# Q1. Create a Pandas Series from a list of integers.
# *Hint:** Use `pd.Series([...])`. The default index starts from 0.

import pandas as pd

S = pd.Series([10,20,30,40,50])
print(S)
```

```python
# Q2. Create a Series from a dictionary with custom index.
# Hint:** Pass a `dict` to `pd.Series()`. Keys become the index labels automatically.

data = {'a': 10, 'b': 20, 'c': 30 }
S = pd.Series(data)
print(S)
```

```python
# Q3. Access elements using index labels.
# Hint:** Use `s["label"]` for a single value, 
# `s[["a","b"]]` for multiple, and `s["a":"c"]` for a range. 
# Label slices are **inclusive** on both ends.

s = pd.Series([10,20,30,40], index=['a', 'b', 'c', 'd'])
print(s['b'])                   # Single element
print(s[['a','b']])             # Multiple element
print(s['a':'c'])               # Range
```

```python
# Q4. Perform arithmetic operations on a Series.
# Hint:** Use `+`, `-`, `*`, `/`, `//`, `%`, `**` 
# directly on the Series. Operations are **element-wise** and **index-aligned**.

s = pd.Series([10,20,30,40])

print(s + 5)
print(s - 5)
print(s * 5)
print(s / 5)
print(s // 5)
print(s ** 5)
```

```python
# Q5. Find mean, min, max of a Series.
# Hint:** Use `.mean()`, `.min()`, `.max()`, `.sum()`, `.median()`, `.std()` directly on the Series.

s = pd.Series([10, 20, 30, 40, 50])

print('Mean :', s.mean())
print('Min :', s.min())
print('Max :', s.max())
print('Sum :', s.sum())
print('Median :', s.median())
print('Std :', s.std())
```

```python
# Q6. Apply conditional filtering on a Series.
# Hint:** Use a boolean mask: `s[s > 50]`. 
# You can also chain conditions with `&` and `|`.

s = pd.Series([10, 20, 30, 40, 50, 60, 70, 80])

print(s [s > 30]) 
print(s [s % 2 == 0])
print(s [(s > 20) & (s < 60)])
print(s[(s > 30) | (s < 50)])
```

```python
# Q7. Replace values in a Series.
# Hint:** Use `.replace(old_val, new_val)` or pass a dictionary `
# {old: new}` to replace multiple values at once.

s = pd.Series([10, 20, 30, 20, 10, 50])

print(s.replace(10, 99))              # Single Replace
print(s.replace({10 : 99, 20 : 88}))  # Multiple replace using dict
print(s.replace([10, 20], [99, 88]))  # Multiple replace using list
```

```python
# Q8. Convert Series data type.
# Hint:** Use `.astype(dtype)` — e.g., `.astype(int)`, `.astype(float)`, `.astype("category")`.

s = pd.Series(['10', '20', '30', '40', '50'])

print(s.astype(int))
print(s.astype(float))
print(s.astype('category'))
```

```python
# Q9. Check for null values in Series.
# Hint:** Use `.isnull()` (returns boolean Series) or `.
# isnull().sum()` for a count.

import pandas as pd
import numpy as np

s = pd.Series([10, 20, 30, np.nan, 40, 50, None])

print(s.isnull())         # Boolean mask = True where null
print(s.isnull().sum())   # Count of null values
print(s.notnull())        # True where Not Null
```

```python
# Q10. Fill missing values in Series.
# Hint:** Use `.fillna(value)` for constant, `
# .fillna(s.mean())` for mean, `.ffill()` for forward fill, `
# .bfill()` for backward fill.

s = pd.Series([10, np.nan, 30, 40, np.nan, 60, 70])

print(s.fillna(1))          # Fill with given number
print(s.fillna(s.mean()))   # Fill with mean
print(s.ffill())            # Forward Fill - Copy previous value
print(s.bfill())            # Backward Fill - Copy next value
```

```python
# Q56. Detect outliers in a column.
# Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1.
#  Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`.
#  Alternatively use Z-score: `|z| > 3`.

df = pd.DataFrame({
    'Salary': [30000, 35000, 40000, 45000, 50000, 60000, 500000]})

Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1

lower_value = Q1 - 1.5 * IQR
upper_value = Q3 + 1.5 * IQR

outliers = df [
    (df['Salary'] < lower_value) |
    (df['Salary'] > upper_value)]
print(outliers)
```

```python
# Q74. Scale numerical data (Z-score standardization).
# Hint:** `(df["col"] - df["col"].mean()) / df["col"].std()` — result has mean=0 and std=1.

df = pd.DataFrame({
    'Age': [20, 30, 40, 50]})

df['Age_ZScore'] = (
    (df['Age'] - df['Age'].mean()) /
    df['Age'].std()
)
print(df)
```

```python
# Q81. Iterate over rows using `iterrows()`.
# Hint:** `for idx, row in df.iterrows():` — `idx` is the index label, `row` is a Series with column values.
# iterrows() loops through each row of the DataFrame.

df = pd.DataFrame({
    'Name' : ['Snehal', 'Komal', 'SK'],
    'Age' : [23, 24, 25]
})

for idx, row in df.iterrows():
    print(idx, row['Name'], row['Age'])
```

```python
# Q82. Iterate using `itertuples()`.
# Hint:** `for row in df.itertuples():` — `row.col_name` accesses values. **Faster** than `iterrows()` due to named tuples.

df = pd.DataFrame({
    'Name' : ['Snehal', 'Komal', 'SK'],
    'Age' : [23, 24, 25]
})
for row in df.itertuples():
    print(row.Name, row.Age)

# itertuples() is a faster alternative to iterrows().
# Instead of returning Series objects, it returns named tuples, 
# which consume less memory and execute faster.
```

```python
# Q83. Iterate over columns.
# Hint:** `for col in df.columns:` then access `df[col]`. Or use `df.items()` which yields `(col_name, Series)` pairs.

# for col in df.columns:
#     print(col)

for col_name, data in df.items(): #It returns (column_name, Series) pairs.
    print(col_name)
    print(data)
```

```python
# Q84. Avoid iteration using vectorization.
# Hint:** Replace `for` loops with direct column operations: 
# `df["new"] = df["a"] * 2 + 1` instead of a loop over rows.

result = []
for value in df['Age']:
    result.append(value * 2 + 1)
df['New Age'] = result
print(result)
```

```python
# Q85. Group data using `groupby()`.
# Hint:** `df.groupby("col")["value_col"].mean()` — specify the grouping key and the aggregation in one line.

df = pd.DataFrame({
    'Department' : ['IT', 'IT', 'HR', 'HR'],
    'Salary' : [50000, 60000, 40000, 45000]
})
print(df.groupby('Department')['Salary'].mean())

```

```python
# Q86. Calculate sum per group.
# Hint:** `df.groupby("col")["value"].sum()` — replace `sum` with any aggregation function.

df.groupby('Department')['Salary'].sum()
```

```python
# Q87. Calculate multiple aggregations.
# Hint:** `df.groupby("col").agg({"col1": "mean", "col2": "sum"})` or `df.groupby("col").agg(name=("col","func"))` for named aggregation.

df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR'],
    'Salary': [50000, 60000, 40000, 45000],
    'Bonus': [5000, 6000, 4000, 4500]
})

result = df.groupby('Department').agg({
    'Salary': 'mean',
    'Bonus': 'sum'
})

print(result)
```

```python
# Q88. Group by multiple columns.
# Hint:** `df.groupby(["col1","col2"])["value"].mean()` — pass a list of column names.

df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR'],
    'Salary': [50000, 60000, 40000, 45000],
    'Bonus': [5000, 6000, 4000, 4500],
    'Gender': ['M', 'F', 'F', 'M'],
})
Groupby =df.groupby(['Department', 'Gender'])['Salary'].mean()
print(Groupby)
```

```python
# Q89. Filter grouped data.
# Hint:** `df.groupby("col").filter(lambda g: g["value"].mean() > threshold)` — keeps entire groups satisfying the condition.
# It keeps only groups that satisfy a condition.

result = df.groupby('Department').filter(lambda x: x['Salary'].mean() > 50000)
print(result)
```

```python
# Q90. Transform grouped data using `transform()`.
# Hint:** `df.groupby("col")["value"].transform("mean")` — returns a Series **same size as the original DataFrame** filled with group statistics.
# transform() : It returns group statistics while keeping the original DataFrame size unchanged.

df['Dept_Avg_Salary'] = (
    df.groupby('Department')['Salary']
      .transform('mean')
)
print(df)
```

```python
# Q91. Apply custom function in groupby.
# Hint:** `df.groupby("col")["value"].agg(lambda s: s.max() - s.min())` — pass any function that takes a Series and returns a scalar.

result = df.groupby('Department')['Salary'].agg(
    lambda s: s.max() - s.min())
print(result)
```

```python
# Q92. Create basic pivot table.
# Hint:** `pd.pivot_table(df, values="val", index="row_col", columns="col_col", aggfunc="mean")`.

pivot = pd.pivot_table(df,
    values='Salary',
    index='Department',
    columns='Gender',
    aggfunc='mean'
)
print(pivot)
```

```python
# Q93. Use multiple aggregation functions in pivot.
# Hint:** Pass a list to `aggfunc`: `aggfunc=["mean","max"]` — the result will have a MultiLevel column header.

pivot = pd.pivot_table(df,
    values='Salary',
    index='Department',
    aggfunc=['mean', 'max'])

print(pivot)
```

```python
# Q94. Handle missing values in pivot tables.
# Hint:** Use `fill_value=0` in `pivot_table()` to replace NaN with a specific value.

pivot = pd.pivot_table(df,
    values='Salary',
    index='Department',
    columns='Gender',
    aggfunc='mean',
    fill_value=0
)
print(pivot)
```

```python
# Q95. Add margins (totals) in pivot tables.
# Hint:** `margins=True, margins_name="TOTAL"` in `pivot_table()` adds a summary row and column.

pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    columns='Gender',
    aggfunc='sum',
    margins=True,
    margins_name='TOTAL'
)
print(pivot)

```

```python
# Q96. Create multi-level pivot table.
# Hint:** Pass lists to `index` and `columns` parameters: `index=["col1","col2"], columns=["col3"]`.

pivot = pd.pivot_table(
    df,
    values='Salary',
    index=['Department', 'Gender'],
    aggfunc='mean'
)

print(pivot)
```

```python
# Q97. Convert DataFrame to pivot format.
# Hint:** `df.pivot(index="row_key", columns="col_key", values="value")` — requires **unique** combinations of row/column keys (no aggregation).

result = df.pivot(
    index='Department',
    columns='Gender',
    values='Salary')
print(result)
```

```python
# Q98. Compare pivot vs groupby.
# Hint:** `groupby()` is more flexible and returns a flat structure. `pivot_table()` gives a 2D grid and handles duplicates with aggregation. `pivot()` does **no aggregation** and fails on duplicates.

df.groupby('Department')['Salary'].mean()
```

```python
# Q99. Build summary report using pivot table.
# Hint:** Combine `pivot_table()` with `margins=True` and `fill_value=0`. Chain `.round(2)` for presentation. Export with `.to_csv()` or `.to_excel()`.

report = pd.pivot_table(df,
    values='Salary',
    index='Department',
    columns='Gender',
    aggfunc='sum',
    fill_value=0,
    margins=True
).round(2)
print(report)
```
