### Q66 · Binary Cross-Entropy loss

```python
# Clipping predictions avoids log(0); the formula averages per-sample losses
y_true = np.array([1,0,1,1,0,0,1,0])
y_pred = np.array([0.9,0.2,0.8,0.7,0.3,0.4,0.85,0.1])
y_pred = np.clip(y_pred, 1e-7, 1-1e-7)
bce = -np.mean(y_true * np.log(y_pred) + (1-y_true) * np.log(1-y_pred))
print("BCE loss:", round(bce, 6))
```

### Q99 · Double normalisation: min-max then Z-score

```python
# Min-max first compresses wildly different scales to [0,1]; Z-score then standardises spread
X = np.random.rand(50, 4) * np.array([1, 100, 1000, 0.01])

col_min = X.min(axis=0)
col_max = X.max(axis=0)
col_range = col_max - col_min
safe_range = np.where(col_range == 0, 1, col_range)   # guard against zero-range column
X_mm = (X - col_min) / safe_range

col_mean = X_mm.mean(axis=0)
col_std  = X_mm.std(axis=0)
safe_std = np.where(col_std == 0, 1, col_std)         # guard against zero-std column
X_z = (X_mm - col_mean) / safe_std

print("Column means ≈ 0:", np.allclose(X_z.mean(axis=0), 0))
print("Column stds ≈ 1: ", np.allclose(X_z.std(axis=0),  1))
```

### Q100 · Full mini ML pipeline (logistic regression, 1 step)

```python
# One gradient-descent step on BCE loss demonstrates the full forward/backward/update cycle
np.random.seed(42)
X = np.random.rand(200, 5)
y = ((X[:,0] + X[:,1]) > 1.0).astype(float)

# 1. Z-score standardise
X = (X - X.mean(axis=0)) / X.std(axis=0)

# 2. Shuffle and split 70/15/15
idx = np.argsort(np.random.rand(200))
X, y = X[idx], y[idx]
X_train, y_train = X[:140], y[:140]
X_val,   y_val   = X[140:170], y[140:170]
X_test,  y_test  = X[170:], y[170:]

# 3. Initialise weights
w = np.zeros(5); b = 0.0

# 4. Forward pass
z     = X_train @ w + b
y_hat = 1 / (1 + np.exp(-z))

# 5. BCE loss (initial)
yh = np.clip(y_hat, 1e-7, 1-1e-7)
loss_init = -np.mean(y_train * np.log(yh) + (1-y_train) * np.log(1-yh))

# 6. Gradient
grad_w = X_train.T @ (y_hat - y_train) / len(y_train)

# 7. Update
w -= 0.1 * grad_w

# 8. New forward pass and loss
z2     = X_train @ w + b
y_hat2 = 1 / (1 + np.exp(-z2))
yh2    = np.clip(y_hat2, 1e-7, 1-1e-7)
loss_new = -np.mean(y_train * np.log(yh2) + (1-y_train) * np.log(1-yh2))

# 9. Report
grad_norm = np.sqrt((grad_w**2).sum())
print(f"Initial loss:  {loss_init:.6f}")
print(f"Updated loss:  {loss_new:.6f}  (should be lower ✓)")
print(f"Gradient norm: {grad_norm:.6f}")
```
