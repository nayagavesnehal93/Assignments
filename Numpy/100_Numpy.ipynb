````python
# 79. Consider 2 sets of points P0,P1 describing lines (2d) and a set of points P, how to compute distance from each point j (P[j]) to each line i (P0[i],P1[i])?
# `No hints provided...`

import numpy as np

# Lines defined by P0 and P1
P0 = np.array([[0, 0],
               [1, 1]])
P1 = np.array([[1, 0],
               [2, 2]])
P = np.array([[1, 2],
              [3, 1],
              [0, 1]])

for point in P:
    for p0, p1 in zip(P0, P1):
        distance = np.abs(np.cross(p1 - p0, point - p0)) / np.linalg.norm(p1 - p0)
        print(distance)
````

````python
# 80. Consider an arbitrary array, write a function that extracts a subpart 
# with a fixed shape and centered on a given element (pad with a `fill` value when necessary)
# `hint: minimum maximum`

import numpy as np

def extract_subarray(arr, center, shape, fill=0):
    
    result = np.full(shape, fill)
    r, c = center
    h, w = shape

    r1 = max(r - h//2, 0)
    r2 = min(r + h//2 + 1, arr.shape[0])

    c1 = max(c - w//2, 0)
    c2 = min(c + w//2 + 1, arr.shape[1])

    sub = arr[r1:r2, c1:c2]
    result[:sub.shape[0], :sub.shape[1]] = sub
    return result
arr = np.arange(25).reshape(5,5)
print(extract_subarray(arr, center=(0,0), shape=(3,3), fill=-1))
````

````python
# 85. Create a 2D array subclass such that Z[i,j] == Z[j,i]
# `hint: class method`

import numpy as np
class SymmetricArray(np.ndarray):

    def __new__(cls, input_array):
        obj = np.asarray(input_array).view(cls)
        return obj

    def __getitem__(self, index):
        i, j = index
        return super().__getitem__((min(i, j), max(i, j)))
Z = SymmetricArray([[1, 2, 3],
                    [2, 4, 5],
                    [3, 5, 6]])

print(Z[0, 2])
print(Z[2, 0])  
````

````python
# 87. Consider a 16x16 array, how to get the block-sum (block size is 4x4)?
# `hint: np.add.reduceat, from numpy.lib.stride_tricks import sliding_window_view (np>=1.20.0)`

import numpy as np

Z = np.arange(256).reshape(16, 16)
block_sum = np.add.reduceat(
    np.add.reduceat(Z, np.arange(0, 16, 4), axis=0),
    np.arange(0, 16, 4),
    axis=1
)
print(block_sum)
````

````python
# 88. How to implement the Game of Life using numpy arrays?
# `No hints provided...`

import numpy as np
def game_of_life(Z):
    N = (
        np.roll(Z, 1, 0) + np.roll(Z, -1, 0) +
        np.roll(Z, 1, 1) + np.roll(Z, -1, 1) +
        np.roll(np.roll(Z, 1, 0), 1, 1) +
        np.roll(np.roll(Z, 1, 0), -1, 1) +
        np.roll(np.roll(Z, -1, 0), 1, 1) +
        np.roll(np.roll(Z, -1, 0), -1, 1)
    )

    return ((N == 3) | ((Z == 1) & (N == 2))).astype(int)
Z = np.random.randint(0, 2, (5, 5))
print("Current Generation:")
print(Z)

Z = game_of_life(Z)
print("\nNext Generation:")
print(Z)
````

````python
# 93. Consider two arrays A and B of shape (8,3) and (2,2). How to find rows of A that contain elements of each row of B regardless of the order of the elements in B?
# `hint: np.where`

import numpy as np

A = np.array([[1, 2, 3],
              [4, 5, 6],
              [1, 4, 2],
              [7, 8, 9],
              [2, 3, 5],
              [4, 1, 5],
              [6, 2, 1],
              [3, 7, 2]])

B = np.array([[1, 2],
              [4, 5]])

result = []
for b in B:
    rows = np.where(np.isin(A, b).sum(axis=1) == len(b))[0]
    result.append(rows)
print(result)
````

````python
# 100. Compute bootstrapped 95% confidence intervals for the mean of a 1D array X (i.e., resample the elements of an array with replacement N times, compute the mean of each sample, and then compute percentiles over the means).
# `hint: np.percentile`

import numpy as np

X = np.array([10, 20, 30, 40, 50])
N = 10000  # Number of bootstrap samples
means = []
for _ in range(N):
    sample = np.random.choice(X, size=len(X), replace=True)
    means.append(sample.mean())
lower, upper = np.percentile(means, [2.5, 97.5])
print("95% Confidence Interval:")
print(lower, upper)
````
