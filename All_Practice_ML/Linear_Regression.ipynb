```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('E:/BRAINWORk/Machine Learning/placement-project/DataSets/placement_n.csv')
```

```python
class MeraLR:
    def __init__(self):
        self.m = None
        self.b = None
        
    def fit(self, X_train, y_train):
        num = 0
        den = 0

        for i in range(X_train.shape[0]):
         num = num + ((X_train[i] - X_train.mean()) * (y_train[i] - y_train.mean()))
         den = den + ((X_train[i] - X_train.mean()) ** 2)      

        self.m = num /den
        self.b = y_train.mean() - (self.m * X_train.mean())
        print(self.m)
        print(self.b)
        
    def predict(self,X_test):

        print(X_test)
        return self.m * X_test + self.b
```
